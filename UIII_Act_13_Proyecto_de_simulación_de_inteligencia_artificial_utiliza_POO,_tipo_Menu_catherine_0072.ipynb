```python
# Programa: Sistema de Criptomonedas
# Programación Orientada a Objetos (POO)

class Criptomonedas_moneda_digital:

    # Constructor
    def __init__(self):

        # Lista para guardar criptomonedas
        self.criptomonedas = []

    # Función para agregar criptomonedas
    def agregar_criptomoneda(self):

        print("\n--- AGREGAR CRIPTOMONEDA ---")

        nombre = input("Nombre de la criptomoneda: ")
        simbolo = input("Símbolo: ")
        blockchain = input("Blockchain: ")
        precio = input("Precio en USD: ")
        mineria = input("Tipo de minería: ")

        crypto = {
            "nombre": nombre,
            "simbolo": simbolo,
            "blockchain": blockchain,
            "precio": precio,
            "mineria": mineria
        }

        self.criptomonedas.append(crypto)

        print("\n✅ Criptomoneda agregada correctamente.")

    # Función para listar criptomonedas
    def listar_criptomoneda(self):

        print("\n================ LISTA DE CRIPTOMONEDAS ================\n")

        if len(self.criptomonedas) == 0:

            print("No hay criptomonedas registradas.")

        else:

            print("{:<5} {:<15} {:<10} {:<15} {:<12} {:<15}".format(
                "ID", "NOMBRE", "SIMBOLO", "BLOCKCHAIN", "PRECIO", "MINERIA"
            ))

            print("-" * 75)

            contador = 0

            for crypto in self.criptomonedas:

                print("{:<5} {:<15} {:<10} {:<15} {:<12} {:<15}".format(
                    contador,
                    crypto["nombre"],
                    crypto["simbolo"],
                    crypto["blockchain"],
                    crypto["precio"],
                    crypto["mineria"]
                ))

                contador += 1

    # Función para actualizar criptomoneda
    def actualizar_criptomoneda(self):

        print("\n--- ACTUALIZAR CRIPTOMONEDA ---")

        self.listar_criptomoneda()

        if len(self.criptomonedas) > 0:

            posicion = int(input("\nIngrese el ID de la criptomoneda: "))

            if posicion < len(self.criptomonedas):

                print("\nIngrese los nuevos datos")

                self.criptomonedas[posicion]["nombre"] = input("Nuevo nombre: ")
                self.criptomonedas[posicion]["simbolo"] = input("Nuevo símbolo: ")
                self.criptomonedas[posicion]["blockchain"] = input("Nueva blockchain: ")
                self.criptomonedas[posicion]["precio"] = input("Nuevo precio USD: ")
                self.criptomonedas[posicion]["mineria"] = input("Nueva minería: ")

                print("\n✅ Criptomoneda actualizada.")

            else:

                print("\n❌ ID no encontrado.")

    # Función para borrar criptomoneda
    def borrar_criptomoneda(self):

        print("\n--- BORRAR CRIPTOMONEDA ---")

        self.listar_criptomoneda()

        if len(self.criptomonedas) > 0:

            posicion = int(input("\nIngrese el ID a eliminar: "))

            if posicion < len(self.criptomonedas):

                self.criptomonedas.pop(posicion)

                print("\n🗑️ Criptomoneda eliminada correctamente.")

            else:

                print("\n❌ ID no válido.")

    # Menú dentro de la clase
    def menu(self):

        while True:

            print("\n")
            print("╔════════════════════════════════════╗")
            print("║     SISTEMA DE CRIPTOMONEDAS 💰   ║")
            print("╠════════════════════════════════════╣")
            print("║ 1. Agregar criptomoneda           ║")
            print("║ 2. Listar criptomonedas           ║")
            print("║ 3. Actualizar criptomoneda        ║")
            print("║ 4. Borrar criptomoneda            ║")
            print("║ 5. Salir                          ║")
            print("╚════════════════════════════════════╝")

            opcion = input("Seleccione una opción: ")

            if opcion == "1":

                self.agregar_criptomoneda()

            elif opcion == "2":

                self.listar_criptomoneda()

            elif opcion == "3":

                self.actualizar_criptomoneda()

            elif opcion == "4":

                self.borrar_criptomoneda()

            elif opcion == "5":

                print("\n👋 Saliendo del sistema...")
                break

            else:

                print("\n❌ Opción incorrecta.")


# Crear objeto
sistema = Criptomonedas_moneda_digital()

# Ejecutar menú
sistema.menu()
```
