```python
# =========================================
# Simulador IA de Criptomonedas
# Todo dentro de la clase
# =========================================

class Criptomonedas_moneda_digital:

    # FUNCION 1
    def crear_datos(self):

        print("========================================")
        print("   IA DE CRIPTOMONEDAS DIGITALES")
        print("========================================")

        self.nombre = input("Nombre de la criptomoneda: ")
        self.precio = float(input("Precio en USD: "))
        self.blockchain = input("Blockchain utilizada: ")
        self.popularidad = input("Popularidad (Alta, Media o Baja): ")
        self.mineria = input("Tipo de minería: ")

    # FUNCION 2
    def mostrar_datos(self):

        print("\n==============================================")
        print("      DATOS DE LA CRIPTOMONEDA")
        print("==============================================")
        print("Nombre:", self.nombre)
        print("Precio:", self.precio, "USD")
        print("Blockchain:", self.blockchain)
        print("Popularidad:", self.popularidad)
        print("Minería:", self.mineria)

    # FUNCION 3
    def subir_precio(self):

        aumento = float(input("¿Cuánto aumentará el precio?: "))
        self.precio = self.precio + aumento

        print("Nuevo precio:", self.precio, "USD")

    # FUNCION 4
    def bajar_precio(self):

        baja = float(input("¿Cuánto bajará el precio?: "))
        self.precio = self.precio - baja

        print("Nuevo precio:", self.precio, "USD")

    # FUNCION 5
    def mostrar_tabla(self):

        print("\n===================================================================")
        print("                 TABLA DE CRIPTOMONEDAS")
        print("===================================================================")

        print(f"{'NOMBRE':15} {'PRECIO':10} {'BLOCKCHAIN':15} {'POPULARIDAD':15} {'MINERIA'}")

        print("-------------------------------------------------------------------")

        print(f"{self.nombre:15} {self.precio:<10} {self.blockchain:15} {self.popularidad:15} {self.mineria}")

        print("===================================================================")

    # MENU
    def menu(self):

        while True:

            print("\n============== MENU ==============")
            print("1. Mostrar datos")
            print("2. Subir precio")
            print("3. Bajar precio")
            print("4. Mostrar tabla")
            print("5. Salir")
            print("==================================")

            opcion = input("Selecciona una opción: ")

            if opcion == "1":
                self.mostrar_datos()

            elif opcion == "2":
                self.subir_precio()

            elif opcion == "3":
                self.bajar_precio()

            elif opcion == "4":
                self.mostrar_tabla()

            elif opcion == "5":
                print("Gracias por usar el simulador IA.")
                break

            else:
                print("Opción incorrecta.")


# =========================================
# CREAR OBJETO
# =========================================

cripto = Criptomonedas_moneda_digital()

cripto.crear_datos()
cripto.menu()
```
